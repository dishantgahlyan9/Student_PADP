# Data preprocessing 

```python
from sklearn.preprocessing import StandardScaler, OneHotEncoder, MinMaxScaler
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.model_selection import StratifiedShuffleSplit , RepeatedStratifiedKFold, RandomizedSearchCV, StratifiedKFold
from sklearn.feature_selection import RFECV
from sklearn.ensemble import RandomForestClassifier
```

```python
# ==============================================================================
# REPLACE CELL 16: Leak-Free Feature Engineering & Ordinal Mapping
# ==============================================================================
# 1. Capture missing value pattern per student BEFORE imputation
df['Null_Count'] = df[['Family_Income', 'Study_Hours_per_Day', 'Stress_Index']].isnull().sum(axis=1)

# 2. Convert 'Semester' ('Year 1', 'Year 2', etc.) into an ordered number (1, 2, 3, 4)
df['Semester_Num'] = df['Semester'].str.replace('Year ', '', regex=False).astype(int)

# 3. Use temporary median-filled series so ratio math doesn't create extra NaNs
study_temp = df['Study_Hours_per_Day'].fillna(df['Study_Hours_per_Day'].median())
stress_temp = df['Stress_Index'].fillna(df['Stress_Index'].median())

# 4. Academic trajectory & compound risk interactions
df['GPA_Trend'] = df['Semester_GPA'] - df['GPA']
df['CGPA_Diff'] = df['Semester_GPA'] - df['CGPA']
df['Study_vs_Travel'] = study_temp / (df['Travel_Time_Minutes'] / 60.0 + 0.1)
df['Attendance_Disruption'] = (100.0 - df['Attendance_Rate']) * df['Assignment_Delay_Days']
df['Stress_Study_Ratio'] = stress_temp / (study_temp + 0.1)
df['Academic_Risk_Index'] = (4.0 - df['GPA']) * (100.0 - df['Attendance_Rate'])

# 5. Update feature lists (Remove 'Semester' from categorical since it is now numerical)
updated_num_features = numerical_features + [
    'Null_Count', 'Semester_Num', 'GPA_Trend', 'CGPA_Diff',
    'Study_vs_Travel', 'Attendance_Disruption', 'Stress_Study_Ratio', 'Academic_Risk_Index'
]
updated_cat_features = ['Gender', 'Internet_Access', 'Part_Time_Job', 'Scholarship', 'Department']

print(f"Total Numerical Features: {len(updated_num_features)}")
print(f"Total Categorical Features: {len(updated_cat_features)}")
```

# Data Spliting 

```python
split = StratifiedShuffleSplit(test_size=0.2, random_state = 42, n_splits=1)
for train_index, test_index in split.split(df, df['Dropout']):
    train_dataset , train_label = df.iloc[train_index], df["Dropout"].iloc[train_index]
    test_dataset, test_label = df.iloc[test_index], df["Dropout"].iloc[test_index]
print(f"The size of train_dataset and train_label is {train_dataset.shape, train_label.shape}")    
print(f"The size of test_dataset and test_label  is {test_dataset.shape, test_label.shape}")
```

```python
type(train_dataset), type(train_label)
```

```python
train_dataset = train_dataset.drop(columns=["Dropout"])
test_dataset = test_dataset.drop(columns = ["Dropout"])
print(train_dataset.shape , test_dataset.shape)
```

# Data Imputation and Standardization

```python
# ==============================================================================
# REPLACE CELL 20 & CELL 22: Imputation, Scaling & Encoding
# ==============================================================================
num_pipeline = Pipeline([
    ("imputation", SimpleImputer(strategy="median")),
    ("Scaling", MinMaxScaler())
])

cat_pipeline = Pipeline([
    ("cat_scaling", OneHotEncoder(handle_unknown="ignore", sparse_output=False, drop="if_binary"))
])

column_transformation = ColumnTransformer([
    ("numerical_transform", num_pipeline, updated_num_features),
    ("categorical_transform", cat_pipeline, updated_cat_features)
])

column_transformation.set_output(transform="pandas")

train_dataset1 = column_transformation.fit_transform(train_dataset)
test_dataset1 = column_transformation.transform(test_dataset)

print(f"Preprocessed Train Shape: {train_dataset1.shape}")
print(f"Preprocessed Test Shape:  {test_dataset1.shape}")
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix,
    ConfusionMatrixDisplay
)
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import (
    RandomForestClassifier,
    GradientBoostingClassifier,
    AdaBoostClassifier
)
from xgboost import XGBClassifier
from lightgbm import LGBMClassifier
from catboost import CatBoostClassifier
```

# Establishing Baseline Model

```python
models = {
    "logisticRegression": LogisticRegression(random_state=42),
    "Support_Vector_Classifier": SVC( probability=True, random_state=42),
    "DecisionTreeClassifier" : DecisionTreeClassifier(random_state=42),
    "RandomForestClassifier" : RandomForestClassifier(random_state=42),
    "GradientBoostingClassifier" :GradientBoostingClassifier(random_state=42),
    "AdaBoostClassifier" : AdaBoostClassifier(random_state=42),
    "XGBoost" : XGBClassifier(random_state=42),
    "lightgbm" : LGBMClassifier(verbose= -1, random_state=42),
    "catboost" : CatBoostClassifier(random_state=42, verbose=0)
}

def evaluate_predictions(y_true, y_pred, y_prob):
    return {
        "Accuracy": accuracy_score(y_true, y_pred),
        "Precision (Macro)": precision_score(y_true, y_pred, average="macro", zero_division=0),
        "Recall (Macro)": recall_score(y_true, y_pred, average="macro", zero_division=0),
        "F1 (Macro)": f1_score(y_true, y_pred, zero_division=0),
        "ROC-AUC": roc_auc_score(y_true, y_prob) if y_prob is not None else np.nan
    }
    
    
train_results = {}
test_results = {}
trained_models = {}
test_predictions = {}

for name, model in models.items():
    # 1. Fit on training data
    model.fit(train_data_selected, train_label)
    trained_models[name] = model

    # 2. Predict on train set
    train_pred = model.predict(train_data_selected)
    train_prob = model.predict_proba(train_data_selected)[:, 1] if hasattr(model, "predict_proba") else None
    train_results[name] = evaluate_predictions(train_label, train_pred, train_prob)

    # 3. Predict on test set
    test_pred = model.predict(test_data_selected)
    test_prob = model.predict_proba(test_data_selected)[:, 1] if hasattr(model, "predict_proba") else None
    test_results[name] = evaluate_predictions(test_label, test_pred, test_prob)
    test_predictions[name] = (test_pred, test_prob)

# Display comparative metrics
df_train_eval = pd.DataFrame(train_results).T.sort_values(by="F1 (Macro)", ascending=False)
df_test_eval = pd.DataFrame(test_results).T.sort_values(by="F1 (Macro)", ascending=False)

print("=== TRAIN SET PERFORMANCE ===")
print(df_train_eval.round(4))
pd.set_option("display.max_columns", None)
pd.set_option("display.width", 1000)
display(df_test_eval.round(4))
print("\n=== TEST SET PERFORMANCE ===")
print(df_test_eval.round(4))


# ==============================================================================
# 4. CONFUSION MATRIX VISUALIZATION FOR ALL MODELS
# ==============================================================================
fig, axes = plt.subplots(nrows=3, ncols=3, figsize=(20, 10))
axes = axes.ravel()

for idx, (name, (pred, _)) in enumerate(test_predictions.items()):
    cm = confusion_matrix(test_label, pred)
    disp = ConfusionMatrixDisplay(confusion_matrix=cm)
    disp.plot(ax=axes[idx], cmap="Blues", colorbar=False)
    axes[idx].set_title(name)

plt.tight_layout()
plt.show()
```

```python
# ==============================================================================
# NEW CELL 25: 5-Fold Cross-Validated Ensemble (LGBM + XGB + CatBoost + RF)
# ==============================================================================
from lightgbm import LGBMClassifier
from xgboost import XGBClassifier
from catboost import CatBoostClassifier
from sklearn.metrics import f1_score, recall_score, precision_score, classification_report

skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

oof_probs = np.zeros(len(train_label))
test_probs = np.zeros(len(test_label))

y_tr_arr = train_label.values
y_te_arr = test_label.values

for fold, (tr_idx, va_idx) in enumerate(skf.split(train_data_selected, y_tr_arr)):
    X_tr, y_tr = train_data_selected.iloc[tr_idx], y_tr_arr[tr_idx]
    X_va, y_va = train_data_selected.iloc[va_idx], y_tr_arr[va_idx]
    
    # 1. LightGBM (Moderate positive weight = 1.5)
    m_lgb = LGBMClassifier(
        n_estimators=400, learning_rate=0.03, num_leaves=20,
        min_child_samples=25, subsample=0.8, colsample_bytree=0.8,
        scale_pos_weight=1.5, random_state=42 + fold, verbose=-1
    )
    m_lgb.fit(X_tr, y_tr)
    
    # 2. XGBoost (Regularized depth + 1.5x weight)
    m_xgb = XGBClassifier(
        n_estimators=400, learning_rate=0.03, max_depth=4,
        min_child_weight=10, subsample=0.8, colsample_bytree=0.8,
        scale_pos_weight=1.5, eval_metric='logloss', random_state=42 + fold
    )
    m_xgb.fit(X_tr, y_tr)
    
    # 3. CatBoost (Symmetric trees resist tabular overfitting)
    m_cat = CatBoostClassifier(
        iterations=500, learning_rate=0.03, depth=5,
        class_weights=[1.0, 1.5], verbose=0, random_state=42 + fold
    )
    m_cat.fit(X_tr, y_tr)
    
    # 4. Pruned Random Forest
    m_rf = RandomForestClassifier(
        n_estimators=300, max_depth=8, min_samples_leaf=8,
        class_weight={0: 1.0, 1: 1.5}, n_jobs=-1, random_state=42 + fold
    )
    m_rf.fit(X_tr, y_tr)
    
    # Blend validation probabilities
    fold_va_prob = (
        0.30 * m_lgb.predict_proba(X_va)[:, 1] +
        0.30 * m_xgb.predict_proba(X_va)[:, 1] +
        0.25 * m_cat.predict_proba(X_va)[:, 1] +
        0.15 * m_rf.predict_proba(X_va)[:, 1]
    )
    oof_probs[va_idx] = fold_va_prob
    
    # Blend test probabilities across all 5 folds
    fold_te_prob = (
        0.30 * m_lgb.predict_proba(test_data_selected)[:, 1] +
        0.30 * m_xgb.predict_proba(test_data_selected)[:, 1] +
        0.25 * m_cat.predict_proba(test_data_selected)[:, 1] +
        0.15 * m_rf.predict_proba(test_data_selected)[:, 1]
    )
    test_probs += fold_te_prob / skf.n_splits

print("5-Fold Ensemble Training Complete!")
```

```python
# ==============================================================================
# NEW CELL 26: Threshold Optimization & Test Set Evaluation
# ==============================================================================
thresholds = np.arange(0.20, 1.0, 0.01)

best_f1_thresh = 0.50
best_oof_f1 = 0.0

recall_80_thresh = 0.50
best_f1_at_80_recall = 0.0

for t in thresholds:
    preds = (oof_probs >= t).astype(int)
    f1 = f1_score(y_tr_arr, preds)
    rec = recall_score(y_tr_arr, preds)
    
    # Track global maximum F1
    if f1 > best_oof_f1:
        best_oof_f1 = f1
        best_f1_thresh = t
        
    # Track best F1 where Recall is at least 80%
    if rec >= 0.80 and f1 > best_f1_at_80_recall:
        best_f1_at_80_recall = f1
        recall_80_thresh = t

# --- Evaluate Mode 1: Maximum F1 Threshold ---
test_preds_max_f1 = (test_probs >= best_f1_thresh).astype(int)
print(f"================ MODE 1: MAX F1 (Threshold = {best_f1_thresh:.2f}) ================")
print(f"Test Recall (Class 1):    {recall_score(y_te_arr, test_preds_max_f1)*100:.2f}%")
print(f"Test Precision (Class 1): {precision_score(y_te_arr, test_preds_max_f1)*100:.2f}%")
print(f"Test F1-Score (Class 1):  {f1_score(y_te_arr, test_preds_max_f1)*100:.2f}%")
print(f"Test Macro F1:            {f1_score(y_te_arr, test_preds_max_f1, average='macro')*100:.2f}%\n")
print(classification_report(y_te_arr, test_preds_max_f1, digits=4))

# --- Evaluate Mode 2: Aggressive Recall (>= 80%) Threshold ---
test_preds_high_rec = (test_probs >= recall_80_thresh).astype(int)
print(f"============= MODE 2: HIGH RECALL (Threshold = {recall_80_thresh:.2f}) =============")
print(f"Test Recall (Class 1):    {recall_score(y_te_arr, test_preds_high_rec)*100:.2f}%")
print(f"Test Precision (Class 1): {precision_score(y_te_arr, test_preds_high_rec)*100:.2f}%")
print(f"Test F1-Score (Class 1):  {f1_score(y_te_arr, test_preds_high_rec)*100:.2f}%")
print(f"Test Macro F1:            {f1_score(y_te_arr, test_preds_high_rec, average='macro')*100:.2f}%\n")
print(classification_report(y_te_arr, test_preds_high_rec, digits=4))
```
